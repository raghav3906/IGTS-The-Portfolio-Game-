```python
# Question 1

nums = [12, 45, 7, 89, 23, 56]

maximum = nums[0]
minimum = nums[0]

for num in nums:
    if num > maximum:
        maximum = num
    if num < minimum:
        minimum = num

print("Maximum:", maximum)
print("Minimum:", minimum)
```

```python
# Question 2

def is_palindrome(s):
    s = s.lower().replace(" ", "")
    return s == s[::-1]

text = "Never Odd Or Even"

if is_palindrome(text):
    print("Palindrome")
else:
    print("Not a palindrome")
```

```python
# Question 3

nums = [1, 8, 12, 5, 14]
x = 4

found = False

for i in range(len(nums)):
    for j in range(i + 1, len(nums)):
        if abs(nums[i] - nums[j]) == x:
            print("Pair found:", nums[i], nums[j])
            found = True
            break
    if found:
        break
```

```python
# Question 5


import pandas as pd

data = {
    'Ticker': ['RELIANCE', 'TCS', 'INFY', 'HDFCBANK', 'WIPRO'],
    'Price': [2850.5, 3900.0, 1500.75, 1650.0, 420.3],
    'Volume': [1200000, 850000, 1100000, 950000, 700000],
    'Sector': ['Energy', 'IT', 'IT', 'Finance', 'IT']
}

df = pd.DataFrame(data)

# (a)
mean_price = df['Price'].mean()
print("Mean Price:", mean_price)

# (b)
total_it_volume = df[df['Sector'] == 'IT']['Volume'].sum()
print("Total IT Volume:", total_it_volume)

# (c)
highest_stock = df.loc[df['Market_Cap_Proxy'].idxmax(), 'Ticker']
print("Highest Market Cap Proxy Stock:", highest_stock)
```

```python
# Question 6

import numpy as np

scores = np.array([45, 82, 76, 55, 90, 38, 67, 71])

# (a)
print("Mean:", np.mean(scores))
print("Median:", np.median(scores))
print("Standard Deviation:", np.std(scores))

# (b)
normalized = (scores - scores.min()) / (scores.max() - scores.min())
print("Normalized Array:", normalized)

# (c)
print("Indices above mean: ")

for i in range(len(scores)):
    if scores[i]> np.mean(scores):
        print(i)
```

```python
# Question 7

import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

daily_returns = np.random.normal(0.0008, 0.015, 252)

# (a)
prices = 100 * np.cumprod(1 + daily_returns)

# (b)
plt.figure(figsize=(10, 5))
plt.plot(prices)
plt.title("Simulated Stock Price Series")
plt.xlabel("Trading Days")
plt.ylabel("Price")
plt.grid(True)
plt.show()

# (c)
final_price = prices[-1]
peak_price = np.max(prices)

running_max = np.maximum.accumulate(prices)
drawdown = (prices - running_max) / running_max
max_drawdown = np.min(drawdown)

print("Final Price:", final_price)
print("Peak Price:", peak_price)
print("Maximum Drawdown:", max_drawdown)
```

```python
# Question 8 

import numpy as np
import pandas as pd

months = ['Jan','Feb','Mar','Apr','May','Jun',
          'Jul','Aug','Sep','Oct','Nov','Dec']

year1 = [42, 45, 50, 48, 55, 60, 58, 63, 67, 70, 75, 80]
year2 = [52, 49, 58, 61, 66, 71, 68, 74, 79, 83, 88, 95]

difference = np.array(year2) - np.array(year1)


df_revenue = pd.DataFrame(
    {
    'Month': months,
    'Year1_Revenue': year1,
    'Year2_Revenue': year2,
    'Difference': difference
}
)


df_revenue['Color'] = np.where(df_revenue['Difference'] > 0, 'Green', 'Red')

print(df_revenue)
```

```python
# Question 9

import numpy as np
import pandas as pd

df = pd.DataFrame({
    'Day': range(1, 11),
    'Price': [100, 103, np.nan, 108, 107, np.nan, 112, 115, np.nan, 120],
    'Volume': [500, 520, 510, np.nan, 530, 545, np.nan, 560, 570, 580]
})

# (a)
print(df.isnull().sum())

# (b)
df['Price'] = df['Price'].ffill()

volume_mean = df['Volume'].mean()
df['Volume'] = df['Volume'].fillna(volume_mean)

# (c)
df['Return'] = df['Price'].pct_change() * 100

print(df)
```

```python
# Question 10 

import numpy as np
import pandas as pd

np.random.seed(42)

returns = np.random.multivariate_normal(
    mean=[0.0005, 0.0008, 0.0003],
    cov=[[0.0004, 0.0001, 0.00005],
         [0.0001, 0.0006, 0.00008],
         [0.00005, 0.00008, 0.0003]],
    size=200
)

prices = 100 * np.cumprod(1 + returns, axis=0)

# (a)
df_prices = pd.DataFrame(
    prices,
    columns=['Asset_A', 'Asset_B', 'Asset_C']
)

print(df_prices.head())

# (b) 
print("\nLast 5 rows of price data:")
print(df_prices.tail())

# (c)
returns_df = df_prices.pct_change().dropna()

correlation_matrix = returns_df.corr()

print("\nCorrelation Matrix:")
print(correlation_matrix)

# (d)
portfolio = df_prices.mean(axis=1)

df_prices['Portfolio'] = portfolio

print("\nPortfolio Values:")
print(df_prices[['Portfolio']].head())
```
