```python
# Question 1

# (a) If P1 plays Top and P2 plays Right,
# the outcome is (0, 5).
# So Player 1 receives 0.

# (b) If both players play Bottom,
# the outcome is (1, 1).
# So Player 2 receives 1.

# (c) If Player 2 plays Left:
# P1 gets 3 from Top and 5 from Bottom.
# So Bottom gives a higher payoff.

# If Player 2 plays Right:
# P1 gets 0 from Top and 1 from Bottom.
# So Bottom again gives a higher payoff.
```

```python
# Question 2

def dominant_strategy(matrix):

    rows = len(matrix)
    cols = len(matrix[0])

    p1_dominant = None
    p2_dominant = None

    # Check dominant strategy for Player 1
    for i in range(rows):
        dominant = True

        for k in range(rows):
            if i == k:
                continue

            for j in range(cols):
                if matrix[i][j][0] <= matrix[k][j][0]:
                    dominant = False
                    break

            if not dominant:
                break

        if dominant:
            p1_dominant = i
            break

    # Check dominant strategy for Player 2
    for j in range(cols):
        dominant = True

        for l in range(cols):
            if j == l:
                continue

            for i in range(rows):
                if matrix[i][j][1] <= matrix[i][l][1]:
                    dominant = False
                    break

            if not dominant:
                break

        if dominant:
            p2_dominant = j
            break

    # Print results
    if p1_dominant is not None:
        print("Player 1 has a strictly dominant strategy: Row", p1_dominant)
    else:
        print("Player 1 has no strictly dominant strategy.")

    if p2_dominant is not None:
        print("Player 2 has a strictly dominant strategy: Column", p2_dominant)
    else:
        print("Player 2 has no strictly dominant strategy.")


# Advertise / Not Advertise
advertise_matrix = [
    [(6,4), (8,1)],
    [(4,6), (7,3)]
]

print("Advertise / Not Advertise")
dominant_strategy(advertise_matrix)

print()


# Prisoner's Dilemma
pd_matrix = [
    [(-1,-1), (-3,0)],
    [(0,-3), (-2,-2)]
]

print("Prisoner's Dilemma")
dominant_strategy(pd_matrix)

print()


# Stag Hunt
stag_hunt = [
    [(4,4), (0,2)],
    [(2,0), (2,2)]
]

print("Stag Hunt")
dominant_strategy(stag_hunt)
```

```python
# Question 3

def iesds(matrix):

    active_rows = list(range(len(matrix)))
    active_cols = list(range(len(matrix[0])))

    changed = True

    while changed:

        changed = False

        for r1 in active_rows:
            for r2 in active_rows:

                if r1 == r2:
                    continue

                dominated = True

                for c in active_cols:
                    if matrix[r1][c][0] >= matrix[r2][c][0]:
                        dominated = False
                        break

                if dominated:
                    print("Removing Row", r1,
                          "because it is strictly dominated by Row", r2)

                    active_rows.remove(r1)
                    changed = True
                    break

            if changed:
                break

        if changed:
            continue

        for c1 in active_cols:
            for c2 in active_cols:

                if c1 == c2:
                    continue

                dominated = True

                for r in active_rows:
                    if matrix[r][c1][1] >= matrix[r][c2][1]:
                        dominated = False
                        break

                if dominated:
                    print("Removing Column", c1,
                          "because it is strictly dominated by Column", c2)

                    active_cols.remove(c1)
                    changed = True
                    break

            if changed:
                break

    print("Remaining Rows:", active_rows)
    print("Remaining Columns:", active_cols)

    if len(active_rows) == 1 and len(active_cols) == 1:
        r = active_rows[0]
        c = active_cols[0]

        print("Predicted Outcome:", (r, c))
        print("Payoffs:", matrix[r][c])


matrix_3x4 = [

    [(1,1), (-1,2), (5,0), (1,1)],
    [(2,3), (1,2), (3,0), (5,1)],
    [(1,1), (0,5), (-1,7), (0,1)]

]

iesds(matrix_3x4)
```

```python
# Question 4

def find_nash_equilibrium(matrix):

    rows = len(matrix)
    cols = len(matrix[0])

    p1_best = set()
    p2_best = set()

    for j in range(cols):

        max_payoff = max(matrix[i][j][0] for i in range(rows))

        for i in range(rows):
            if matrix[i][j][0] == max_payoff:
                p1_best.add((i, j))

    for i in range(rows):

        max_payoff = max(matrix[i][j][1] for j in range(cols))

        for j in range(cols):
            if matrix[i][j][1] == max_payoff:
                p2_best.add((i, j))

    ne = p1_best.intersection(p2_best)

    if ne:
        print("Pure Strategy Nash Equilibria:")
        for eq in ne:
            print(eq, "with payoff", matrix[eq[0]][eq[1]])
    else:
        print("No Pure Strategy Nash Equilibrium found.")


# Prisoner's Dilemma
print("Prisoner's Dilemma")

pd = [
    [(-1,-1), (-3,0)],
    [(0,-3), (-2,-2)]
]

find_nash_equilibrium(pd)

print()


# Stag Hunt
print("Stag Hunt")

stag = [
    [(4,4), (0,2)],
    [(2,0), (2,2)]
]

find_nash_equilibrium(stag)

print()


# Chicken
print("Chicken")

chicken = [
    [(0,0), (-1,1)],
    [(1,-1), (-10,-10)]
]

find_nash_equilibrium(chicken)

print()


# Battle of the Sexes
print("Battle of the Sexes")

bos = [
    [(2,1), (0,0)],
    [(0,0), (1,2)]
]

find_nash_equilibrium(bos)

print()


# 3x3 Matrix
print("3x3 Matrix")

matrix_3x3 = [
    [(4,4), (0,0), (1,3)],
    [(0,0), (5,5), (0,0)],
    [(3,1), (0,0), (6,6)]
]

find_nash_equilibrium(matrix_3x3)
```

```python
# Question 5

from itertools import product

def n_player_nash(strategies, payoffs):

    num_players = len(strategies)

    all_profiles = list(product(*strategies))

    equilibria = []

    for profile in all_profiles:

        is_ne = True

        for i in range(num_players):

            current_payoff = payoffs[profile][i]

            for alt_strategy in strategies[i]:

                if alt_strategy == profile[i]:
                    continue

                new_profile = list(profile)
                new_profile[i] = alt_strategy
                new_profile = tuple(new_profile)

                if payoffs[new_profile][i] > current_payoff:
                    is_ne = False
                    break

            if not is_ne:
                break

        if is_ne:
            equilibria.append(profile)

    if equilibria:
        print("Nash Equilibria:")
        for eq in equilibria:
            print(eq, "->", payoffs[eq])
    else:
        print("No Pure Strategy Nash Equilibrium found.")


strategies = [['C','D'], ['C','D'], ['C','D']]

payoffs = {

('C','C','C'): [4,4,4],
('C','C','D'): [1,1,6],
('C','D','C'): [1,6,1],
('C','D','D'): [0,3,3],

('D','C','C'): [6,1,1],
('D','C','D'): [3,0,3],
('D','D','C'): [3,3,0],
('D','D','D'): [2,2,2]

}

n_player_nash(strategies, payoffs)

# The Nash Equilibrium is typically ('D','D','D').
# This shows that even though cooperation gives better collective payoff,
# players choose to defect because unilateral deviation is profitable.
# It demonstrates the difficulty of maintaining cooperation in collective action problems.
```

```python
# Question 6

def dominant_strategy(matrix):

    rows = len(matrix)
    cols = len(matrix[0])

    print("\n--- Dominant Strategy Analysis ---")

    found = False

    for i in range(rows):

        dominant = True

        for k in range(rows):

            if i == k:
                continue

            for j in range(cols):

                if matrix[i][j][0] <= matrix[k][j][0]:
                    dominant = False
                    break

            if not dominant:
                break

        if dominant:
            print("Player 1 dominant strategy: Row", i)
            found = True

    if not found:
        print("Player 1 has no dominant strategy.")

    found = False

    for j in range(cols):

        dominant = True

        for l in range(cols):

            if j == l:
                continue

            for i in range(rows):

                if matrix[i][j][1] <= matrix[i][l][1]:
                    dominant = False
                    break

            if not dominant:
                break

        if dominant:
            print("Player 2 dominant strategy: Column", j)
            found = True

    if not found:
        print("Player 2 has no dominant strategy.")


def iesds(matrix):

    print("\n--- IESDS Analysis ---")

    active_rows = list(range(len(matrix)))
    active_cols = list(range(len(matrix[0])))

    changed = True

    while changed:

        changed = False

        
        for r1 in active_rows:
            for r2 in active_rows:

                if r1 == r2:
                    continue

                dominated = True

                for c in active_cols:
                    if matrix[r1][c][0] >= matrix[r2][c][0]:
                        dominated = False
                        break

                if dominated:
                    print("Remove Row", r1)
                    active_rows.remove(r1)
                    changed = True
                    break

            if changed:
                break

        if changed:
            continue

        
        for c1 in active_cols:
            for c2 in active_cols:

                if c1 == c2:
                    continue

                dominated = True

                for r in active_rows:
                    if matrix[r][c1][1] >= matrix[r][c2][1]:
                        dominated = False
                        break

                if dominated:
                    print("Remove Column", c1)
                    active_cols.remove(c1)
                    changed = True
                    break

            if changed:
                break

    print("Remaining Rows:", active_rows)
    print("Remaining Columns:", active_cols)


def find_nash(matrix):

    print("\n--- Nash Equilibrium Analysis ---")

    rows = len(matrix)
    cols = len(matrix[0])

    p1_best = set()
    p2_best = set()

    for j in range(cols):

        best = max(matrix[i][j][0] for i in range(rows))

        for i in range(rows):
            if matrix[i][j][0] == best:
                p1_best.add((i, j))

    for i in range(rows):

        best = max(matrix[i][j][1] for j in range(cols))

        for j in range(cols):
            if matrix[i][j][1] == best:
                p2_best.add((i, j))

    ne = p1_best.intersection(p2_best)

    if ne:
        print("Nash Equilibria:")
        for eq in ne:
            print(eq, matrix[eq[0]][eq[1]])
    else:
        print("No Nash Equilibrium found.")


rows = int(input("Enter number of strategies for Player 1: "))
cols = int(input("Enter number of strategies for Player 2: "))

matrix = []

print("\nEnter payoffs row by row:")

for i in range(rows):

    row = []

    for j in range(cols):

        p1, p2 = map(int, input(
            f"Cell ({i},{j}) -> ").split())

        row.append((p1, p2))

    matrix.append(row)

print("\nPayoff Matrix:")

for row in matrix:
    print(row)


dominant_strategy(matrix)
iesds(matrix)
find_nash(matrix)
```
